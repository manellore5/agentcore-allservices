# Notebook 06: Code Interpreter Integration

## Learning Objectives
- Set up AgentCore Code Interpreter with Strands
- Create travel agent with analytical capabilities
- Test budget calculations and data analysis

## Prerequisites
- AWS credentials configured
- strands-agents-tools package
This notebook runs TypeScript on the Deno kernel. Pick the **Deno** kernel in the top right.


```typescript
// AWS Configuration
Deno.env.set("AWS_REGION", "us-east-1");

// APPROACH A: Use credentials
// Deno.env.set("AWS_ACCESS_KEY_ID", "your_access_key");
// Deno.env.set("AWS_SECRET_ACCESS_KEY", "your_secret_key");
// Deno.env.set("AWS_SESSION_TOKEN", "your_session_token");

// APPROACH B: Use AWS SSO profile
// Deno.env.set("AWS_PROFILE", "your_profile");

console.log("\u2705 AWS configured");
```

```typescript
// Imports
import { Agent, BedrockModel } from "@strands-agents/sdk";
import { CodeInterpreterTools } from "bedrock-agentcore/experimental/code-interpreter/strands";
import { loadEnv, writeFile } from "../shared/notebook.ts";

await loadEnv();

console.log("\u2705 Imports successful");
```

```typescript
// Create tools and agent
const codeInterpreter = new CodeInterpreterTools({ region: "us-east-1" });
const model = new BedrockModel({ modelId: "us.anthropic.claude-sonnet-4-6" });

const agent = new Agent({
  model,
  tools: [...codeInterpreter.tools],
  systemPrompt: `You are a travel budget analyst. Use the code interpreter to run Python for
calculations, comparisons and statistics. Always show the numbers you computed.`,
});

console.log(`\u2705 Agent created with ${codeInterpreter.tools.length} code interpreter tools`);
```

```typescript
// Test 1: Budget calculation
const response1 = await agent.invoke(
  "Calculate a budget breakdown for a $3000, 7-day trip to Italy. Show amounts for flights, hotels, food, and activities.",
);
console.log("Budget Analysis:");
console.log(response1.toString());
```

```typescript
// Test 2: Cost comparison
const response2 = await agent.invoke(
  "Compare cost per day for these trips: Japan $5000/10 days, Thailand $2000/14 days, France $4000/8 days. Which is most cost-effective?",
);
console.log("Cost Comparison:");
console.log(response2.toString());
```

```typescript
// Test 3: Currency and savings calculation
const response3 = await agent.invoke(
  "If I save $200 per month for travel, how long to reach $4500? Also calculate compound interest at 3% annual rate.",
);
console.log("Savings Analysis:");
console.log(response3.toString());
```

```typescript
// Test 4: Statistical analysis
const response4 = await agent.invoke(
  "Analyze these hotel prices per night: $120, $95, $180, $110, $150, $85, $200. Calculate mean, median, and recommend budget range.",
);
console.log("Price Analysis:");
console.log(response4.toString());
```

```typescript
// Release the sandbox session: it is billed while it lives
await codeInterpreter.stopSession();
console.log("\ud83d\uded1 Code interpreter session stopped");
```

```typescript
// Save setup info
const config = {
  code_interpreter: "configured",
  capabilities: ["budget_analysis", "cost_comparison", "statistical_analysis"],
};

await writeFile("../backend/code_interpreter_config.json", `${JSON.stringify(config, null, 2)}\n`);

console.log("\u2705 Code interpreter configuration saved");
```
